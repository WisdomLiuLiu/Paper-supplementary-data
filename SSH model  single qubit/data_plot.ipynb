```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from qutip import *
from math import *
from scipy.optimize import minimize
import scienceplots
plt.style.use('science')
mpl.rcParams.update({
    "text.usetex": True,
    # "font.size": 10,          # 涓庢枃涓鏂囧瓧鍙蜂竴鑷碶n    # "axes.labelsize": 10,
    # "xtick.labelsize": 8,
    # "ytick.labelsize": 8,
    # "legend.fontsize": 8,
    "text.latex.preamble": r"""
        \usepackage{newtxtext,newtxmath}  % APS 榛樿瀛椾綋
    """
})
```

```python
exp_lchs_40_1 = np.loadtxt('./data_lchs_benchmark_k_dk/40_1.txt')
exp_lchs_40_2 = np.loadtxt('./data_lchs_benchmark_k_dk/40_2.txt')
exp_lchs_80_1 = np.loadtxt('./data_lchs_benchmark_k_dk/80_1.txt')
exp_lchs_80_2 = np.loadtxt('./data_lchs_benchmark_k_dk/80_2.txt')
exp_exact = np.loadtxt('./data_lchs_benchmark_k_dk/exp_exact.txt')

dt = 0.05
time_list = np.arange(0,1.5,dt)

width = 297/25.4
high = width*0.7
fontsize_ = 30

linewidth_ = 3
fig,ax = plt.subplots(1,1,figsize=(width,high))

ax.plot(time_list,exp_exact,label='Theory',linewidth=linewidth_)
ax.plot(time_list,exp_lchs_40_1,'--o',linewidth=linewidth_,label='$K=40,\delta k=1$')
ax.plot(time_list,exp_lchs_40_2,'--^',linewidth=linewidth_,label='$K=40,\delta k=2$')
ax.plot(time_list,exp_lchs_80_1,'--s',linewidth=linewidth_,label='$K=80,\delta k=1$')
ax.plot(time_list,exp_lchs_80_2,'--*',linewidth=linewidth_,label='$K=80,\delta k=2$')

ax.set_ylabel('$\langle P_0 \\rangle$',fontsize=fontsize_)
ax.set_xlabel('Time $t$',fontsize=fontsize_)


ax.set_xticks([0,0.5,1,1.5])
ax.set_yticks([0.6,0.7,0.8,0.9,1.0])
ax.tick_params(axis='both',labelsize=fontsize_,pad=6)
for index in ['left','right','top','bottom']:
    ax.spines[index].set_linewidth(linewidth_)
ax.legend(fontsize=fontsize_)
plt.tight_layout()
plt.savefig('ssh_bench_lchs.pdf')
plt.show()
```

```python
epoch = [x+1 for x in range(20)]

opt_dt = np.loadtxt('dt.txt')

opt_2dt = np.loadtxt('2dt.txt')

width = 297/25.4
high = width*0.4
fontsize_ = 30
linewidth_ = 1.5


fig,ax = plt.subplots(1,2,figsize=(width,high))

ax[0].plot(epoch,opt_dt,'--o',linewidth=linewidth_)
ax[0].set_xlabel('Epoch',fontsize=fontsize_)
ax[0].set_ylabel('Loss',fontsize=fontsize_)
ax[0].set_title('$0 \\to \delta t$',fontsize=fontsize_,pad=6)
ax[0].set_xticks([0,10,20])
ax[0].set_yticks([0,0.05,0.1,0.15])
ax[0].tick_params(axis='both',labelsize=fontsize_,pad=6)
for index in ['left','right','top','bottom']:
    ax[0].spines[index].set_linewidth(linewidth_)


ax[1].plot(epoch,opt_2dt,'--o',linewidth=linewidth_)
ax[1].set_xlabel('Epoch',fontsize=fontsize_)
# ax[1].set_ylabel('Loss',fontsize=fontsize_)
ax[1].set_title('$\delta t \\to 2\delta t$',fontsize=fontsize_,pad=6)
ax[1].set_xticks([0,10,20])
ax[1].set_yticks([-0.005,0,0.005,0.01])
ax[1].tick_params(axis='both',labelsize=fontsize_,pad=6)
for index in ['left','right','top','bottom']:
    ax[1].spines[index].set_linewidth(linewidth_)

fig.text(0.04,0.92,r'\bfseries (a)',fontsize=fontsize_)
fig.text(0.5,0.92,r'\bfseries (b)',fontsize=fontsize_)

plt.tight_layout()
plt.savefig('ssh_optimize.pdf')
plt.show()
```
