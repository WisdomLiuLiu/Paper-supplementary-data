```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from qutip import *
from math import *
from scipy.optimize import minimize
import scienceplots
plt.style.use('science')
mpl.rcParams.update({
    "text.usetex": True,
    # "font.size": 10,          # 涓庢枃涓鏂囧瓧鍙蜂竴鑷碶n    # "axes.labelsize": 10,
    # "xtick.labelsize": 8,
    # "ytick.labelsize": 8,
    # "legend.fontsize": 8,
    "text.latex.preamble": r"""
        \usepackage{newtxtext,newtxmath}  % APS 榛樿瀛椾綋
    """
})
```

```python
exp_05 = np.loadtxt('./data_exact/exp_05.txt')
exp_1 = np.loadtxt('./data_exact/exp_1.txt')
exp_2 = np.loadtxt('./data_exact/exp_2.txt')
exp_exact_05 = exp_05[:50]
exp_exact_1 = exp_1[:50]
exp_exact_2 = exp_2[:50]

exp_vqa_mean_05, exp_vqa_std_05 = [], []
exp_vqa_mean_1, exp_vqa_std_1 = [], []
exp_vqa_mean_2, exp_vqa_std_2 = [], []
for index in range(50):
    exp = np.loadtxt('./exp_05/{}.txt'.format(index))
    exp_vqa_mean_05.append(np.mean(exp))
    exp_vqa_std_05.append(np.std(exp))
    exp = np.loadtxt('./exp_1/{}.txt'.format(index))
    exp_vqa_mean_1.append(np.mean(exp))
    exp_vqa_std_1.append(np.std(exp))
    exp = np.loadtxt('./exp_2/{}.txt'.format(index))
    exp_vqa_mean_2.append(np.mean(exp))
    exp_vqa_std_2.append(np.std(exp))

dt = 0.1
time_list_short = np.arange(0, 5, dt)
width = 297 / 25.4
high = width * 0.8
linewidth_ = 1.5
```

```python

fontsize_ = 20

fig,ax = plt.subplots(1,3,figsize=(width,high*0.4))




ax[0].plot(time_list_short,exp_exact_05,linewidth=linewidth_,label='Theory')
ax[0].errorbar(time_list_short,exp_vqa_mean_05,yerr=exp_vqa_std_05,fmt='--o',linewidth=linewidth_,label='VQA')
# ax[0,1].set_ylabel('$\langle S_z \\rangle$',fontsize=fontsize_)
ax[0].set_ylabel('$\langle S_z \\rangle$',fontsize=fontsize_)
ax[0].set_title('$g_i=0.5$',fontsize=fontsize_,pad=10)
ax[0].set_xlabel('Time $t$',fontsize=fontsize_)
ax[0].set_xticks([0,2.5,5])
ax[0].set_yticks([-1,-0.5,0,0.5,1])
ax[0].tick_params(axis='both',labelsize=fontsize_,pad=10)
for index in ['left','right','top','bottom']:
    ax[0].spines[index].set_linewidth(linewidth_)
# ax[0,1].legend(fontsize=fontsize_)


ax[1].plot(time_list_short,exp_exact_1,linewidth=linewidth_,label='Theory')
ax[1].errorbar(time_list_short,exp_vqa_mean_1,yerr=exp_vqa_std_1,fmt='--o',linewidth=linewidth_,label='VQA')

ax[1].set_title('$g_i=1$',fontsize=fontsize_,pad=10)
ax[1].set_xlabel('Time $t$',fontsize=fontsize_)
ax[1].set_xticks([0,2.5,5])
ax[1].set_yticks([-1,-0.5,0,0.5,1])
ax[1].tick_params(axis='both',labelsize=fontsize_,pad=10)
for index in ['left','right','top','bottom']:
    ax[1].spines[index].set_linewidth(linewidth_)
# ax[1,0].legend(fontsize=fontsize_)


ax[2].plot(time_list_short,exp_exact_2,linewidth=linewidth_,label='Theory')
ax[2].errorbar(time_list_short,exp_vqa_mean_2,yerr=exp_vqa_std_2,fmt='--o',linewidth=linewidth_,label='Experiment')
# ax[1,1].set_ylabel('$\langle S_z \\rangle$',fontsize=fontsize_)
ax[2].set_title('$g_i=2$',fontsize=fontsize_,pad=10)
ax[2].set_xlabel('Time $t$',fontsize=fontsize_)
ax[2].set_xticks([0,2.5,5])
ax[2].set_yticks([-1,-0.5,0,0.5,1])
ax[2].tick_params(axis='both',labelsize=fontsize_,pad=10)
for index in ['left','right','top','bottom']:
    ax[2].spines[index].set_linewidth(linewidth_)
ax[2].legend(fontsize=fontsize_)

fig.text(0.03,0.9,r'\bfseries (a)',fontsize=fontsize_)
fig.text(0.35,0.9,r'\bfseries (b)',fontsize=fontsize_)
fig.text(0.66,0.9,r'\bfseries (c)',fontsize=fontsize_)

plt.tight_layout()
plt.savefig('ising_errorbar.pdf')
plt.show()
```
