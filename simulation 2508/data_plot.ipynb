```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from qutip import *
from math import *
from scipy.optimize import minimize
import scienceplots
plt.style.use('science')
mpl.rcParams.update({
    "text.usetex": True,
    # "font.size": 10,          # 涓庢枃涓鏂囧瓧鍙蜂竴鑷碶n    # "axes.labelsize": 10,
    # "xtick.labelsize": 10,
    # "ytick.labelsize": 10,
    # "legend.fontsize": 10,
    "text.latex.preamble": r"""
        \usepackage{newtxtext,newtxmath}  % APS 榛樿瀛椾綋
    """
})
```

```python
dt = 0.2
time_list = np.arange(0, 4.01, dt)

flist_with = np.loadtxt('supp_0203/fidelity_with.txt')
flist_without = np.loadtxt('supp_0203/fidelity_without.txt')

total_occ_with = np.loadtxt('supp_0203/total_occ_with.txt')
total_occ_without = np.loadtxt('supp_0203/total_occ_without.txt')



fontsize_ = 30
width = 297/25.4
high = width * 0.5
linewidth_ = 1.5
fig,ax = plt.subplots(1,2,figsize=(width,high))



ax[0].plot(time_list,flist_with,'--o',linewidth=linewidth_,label='with penalty')
ax[0].plot(time_list,flist_without,'--o',linewidth=linewidth_,label='without penalty')
ax[0].set_ylabel('fidelity',fontsize=fontsize_)
ax[0].set_title('fidelity compare',fontsize=fontsize_,pad=10)
ax[0].set_xlabel('time $t$',fontsize=fontsize_)
ax[0].set_xticks([0,1,2,3,4])
ax[0].set_yticks([0.90,0.95,1])
ax[0].set_ylim([0.89,1.01])
ax[0].tick_params(axis='both',labelsize=fontsize_,pad=10)
for index in ['left','right','top','bottom']:
    ax[0].spines[index].set_linewidth(linewidth_)
# ax[0].legend(fontsize=fontsize_-5)

ax[1].plot(time_list,total_occ_with,'--o',linewidth=linewidth_,label='with penalty')
ax[1].plot(time_list,total_occ_without,'--o',linewidth=linewidth_,label='without penalty')
ax[1].set_title('particle number compare',fontsize=fontsize_,pad=10)
ax[1].set_xlabel('time $t$',fontsize=fontsize_)
ax[1].set_ylabel('$\sum_j \langle n_j\\rangle$',fontsize=fontsize_)
ax[1].set_xticks([0,1,2,3,4])
ax[1].set_yticks([2,2.05,2.1])
ax[1].set_ylim([1.99,2.11])
ax[1].tick_params(axis='both',labelsize=fontsize_,pad=10)
for index in ['left','right','top','bottom']:
    ax[1].spines[index].set_linewidth(linewidth_)
ax[1].legend(fontsize=fontsize_-5)

fig.text(0.04,0.92,r'\bfseries (a)',fontsize=fontsize_)
fig.text(0.52,0.92,r'\bfseries (b)',fontsize=fontsize_)

plt.tight_layout(w_pad=2)
plt.savefig('hn_penalty_compare.pdf')
plt.show()
```
