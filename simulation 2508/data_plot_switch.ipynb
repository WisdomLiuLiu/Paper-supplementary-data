```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from qutip import *
from math import *
from scipy.optimize import minimize
import scienceplots
plt.style.use('science')
mpl.rcParams.update({
    "text.usetex": True,
    # "font.size": 10,          # 涓庢枃涓鏂囧瓧鍙蜂竴鑷碶n    # "axes.labelsize": 10,
    # "xtick.labelsize": 10,
    # "ytick.labelsize": 10,
    # "legend.fontsize": 10,
    "text.latex.preamble": r"""
        \usepackage{newtxtext,newtxmath}  % APS 榛樿瀛椾綋
    """
})
```

```python
qubit_number = 10
dt = 0.2
time_list = np.arange(0, 4.01, dt)

exp_exact_hp = np.loadtxt('./g1u1/exp_exact_g1u1.txt')
exp_vqa_hp = np.loadtxt('./g1u1/exp_vqa_g1u1.txt')
exp_exact_hp_switch = np.loadtxt('./LR_switch/exp_exact_switch.txt')
exp_vqa_hp_switch = np.loadtxt('./LR_switch/exp_vqa_switch.txt')

exp_exact_hp_z8 = exp_exact_hp[8]
exp_exact_hp_z9 = exp_exact_hp[9]
exp_exact_hp_switch_n8 = exp_exact_hp_switch[8]
exp_exact_hp_switch_n9 = exp_exact_hp_switch[9]
```

```python

fontsize_ = 26
width = 3.4*3
high = width * 0.8
linewidth_ = 1.5

# 鍋囪宸茬粡鏈夎繖浜涙暟鎹細exp_exact_m1, exp_exact_1, exp_vqa_m1, exp_vqa_1
# 杩欓噷鎵惧叏灞€鐨?vmin / vmax
all_data = np.array([
    exp_exact_hp.flatten(),
    exp_exact_hp_switch.flatten(),
    exp_vqa_hp.flatten(),
    exp_vqa_hp_switch.flatten()
])
vmin = all_data.min()
vmax = all_data.max()

fig, ax = plt.subplots(2, 2, figsize=(width, high),constrained_layout=True)
fig.set_constrained_layout_pads(w_pad=0.02,  # 瀹芥柟鍚戠殑棰濆闂磋窛锛堝崟浣嶏細鑻卞锛塡n                                h_pad=0.02,  # 楂樻柟鍚戠殑棰濆闂磋窛
                                hspace=0.05,  # 缃戞牸鍨傜洿棰濆姣斾緥
                                wspace=0.05)  

y_tick = [x+0.5 for x in range(0,qubit_number,2)]
y_tick_label = [qubit_number - 1 - x for x in range(0,qubit_number,2)]
x_tick = [0, 1, 2, 3, 4]

im00 = ax[0,0].imshow(exp_exact_hp, extent=[min(time_list), max(time_list), 0, qubit_number],
                      aspect='auto', vmin=vmin, vmax=vmax)
ax[0,0].set_xticks(ticks=x_tick, labels=x_tick, fontsize=fontsize_)
ax[0,0].set_yticks(ticks=y_tick, labels=y_tick_label, fontsize=fontsize_)
ax[0,0].tick_params(axis='both', pad=6)
ax[0,0].set_ylabel('qubit index $j$', fontsize=fontsize_)
ax[0,0].set_title('Theory, $H_{\\text{p}}(t_R,t_L)$', fontsize=fontsize_, pad=12)

im01 = ax[0,1].imshow(exp_exact_hp_switch, extent=[min(time_list), max(time_list), 0, qubit_number],
                      aspect='auto', vmin=vmin, vmax=vmax)
ax[0,1].set_xticks(ticks=x_tick, labels=x_tick, fontsize=fontsize_)
ax[0,1].set_yticks(ticks=y_tick, labels=y_tick_label, fontsize=fontsize_)
ax[0,1].tick_params(axis='both', pad=6)
ax[0,1].set_title('Theory, $H_{\\text{p}}(t_L,t_R)$', fontsize=fontsize_, pad=12)

im10 = ax[1,0].imshow(exp_vqa_hp, extent=[min(time_list), max(time_list), 0, qubit_number],
                      aspect='auto', vmin=vmin, vmax=vmax)
ax[1,0].set_xticks(ticks=x_tick, labels=x_tick, fontsize=fontsize_)
ax[1,0].set_yticks(ticks=y_tick, labels=y_tick_label, fontsize=fontsize_)
ax[1,0].tick_params(axis='both', pad=6)
ax[1,0].set_xlabel('time $t$', fontsize=fontsize_)
ax[1,0].set_ylabel('qubit index $j$', fontsize=fontsize_)
ax[1,0].set_title('VQS, $H_{\\text{p}}(t_R,t_L)$', fontsize=fontsize_, pad=12)


im11 = ax[1,1].imshow(exp_vqa_hp_switch, extent=[min(time_list), max(time_list), 0, qubit_number],
                      aspect='auto', vmin=vmin, vmax=vmax)
ax[1,1].set_xticks(ticks=x_tick, labels=x_tick, fontsize=fontsize_)
ax[1,1].set_yticks(ticks=y_tick, labels=y_tick_label, fontsize=fontsize_)
ax[1,1].tick_params(axis='both', pad=6)
ax[1,1].set_xlabel('time $t$', fontsize=fontsize_)
ax[1,1].set_title('VQS, $H_{\\text{p}}(t_L,t_R)$', fontsize=fontsize_, pad=12)



# 鍏叡 colorbar锛堢粦瀹氬埌 fig锛塡ncbar = fig.colorbar(im00, ax=ax, orientation='vertical', fraction=0.02, pad=0.06, shrink=0.5)
cbar = fig.colorbar(im00, ax=ax, orientation='vertical', fraction=0.02, pad=0.06, shrink=0.5)
cbar.ax.tick_params(labelsize=fontsize_)
# cbar.set_label("Your Label", fontsize=fontsize_)


fig.text(0.01,0.96,r'\bfseries (a)',fontsize=fontsize_)
fig.text(0.44,0.96,r'\bfseries (b)',fontsize=fontsize_)
fig.text(0.01,0.48,r'\bfseries (c)',fontsize=fontsize_)
fig.text(0.44,0.48,r'\bfseries (d)',fontsize=fontsize_)

# plt.tight_layout()
plt.savefig('hn_num_switch.pdf')
plt.show()
```

```python
mean_lr_z8 = []
std_lr_z8 = []

mean_lr_z9 = []
std_lr_z9 = []

mean_rl_z8 = []
std_rl_z8 = []

mean_rl_z9 = []
std_rl_z9 = []

for index in range(21):
    exp_z8 = np.loadtxt('./g1u1/exp_8_2026/{}.txt'.format(index))
    exp_z9 = np.loadtxt('./g1u1/exp_9_2026/{}.txt'.format(index))
    exp_rl_z8 = np.loadtxt('./LR_switch/exp_8_2026/{}.txt'.format(index))
    exp_rl_z9 = np.loadtxt('./LR_switch/exp_9_2026/{}.txt'.format(index))
    mean_lr_z8.append(np.mean(exp_z8))
    std_lr_z8.append(np.std(exp_z8))
    mean_lr_z9.append(np.mean(exp_z9))
    std_lr_z9.append(np.std(exp_z9))
    mean_rl_z8.append(np.mean(exp_rl_z8))
    std_rl_z8.append(np.std(exp_rl_z8))
    mean_rl_z9.append(np.mean(exp_rl_z9))
    std_rl_z9.append(np.std(exp_rl_z9))



fontsize_ = 30
width = 297/25.4
high = width * 0.5
linewidth_ = 1.5
fig,ax = plt.subplots(1,2,figsize=(width,high))



ax[0].plot(time_list,exp_exact_hp_z8,linewidth=linewidth_,label='The. $q_8$')
ax[0].errorbar(time_list,mean_lr_z8,yerr=std_lr_z8,fmt='--o',linewidth=linewidth_,label='Sim. $q_8$')
# ax[0].plot(time_list,exp_exp_hp_z8,'--o',linewidth=linewidth_,label='Sim. $q_8$')
ax[0].plot(time_list,exp_exact_hp_z9,linewidth=linewidth_,label='The. $q_9$')
# ax[0].plot(time_list,exp_exp_hp_z9,'--o',linewidth=linewidth_,label='Sim. $q_9$')
ax[0].errorbar(time_list,mean_lr_z9,yerr=std_lr_z9,fmt='--o',linewidth=linewidth_,label='Sim. $q_9$')
ax[0].set_ylabel('particle number $\langle n \\rangle$',fontsize=fontsize_)
ax[0].set_title('$H_{\\text{p}}(t_R,t_L)$',fontsize=fontsize_,pad=12)
ax[0].set_xlabel('time $t$',fontsize=fontsize_)
ax[0].set_xticks([0,1,2,3,4])
ax[0].set_yticks([.5,0,0.5,1])
ax[0].tick_params(axis='both',labelsize=fontsize_,pad=10)
for index in ['left','right','top','bottom']:
    ax[0].spines[index].set_linewidth(linewidth_)


ax[1].plot(time_list,exp_exact_hp_switch_n8,linewidth=linewidth_,label='The. $q_8$')
# ax[1].plot(time_list,exp_exp_hp_switch_z8,'--o',linewidth=linewidth_,label='Sim. $q_8$')
ax[1].errorbar(time_list,mean_rl_z8,yerr=std_rl_z8,fmt='--o',linewidth=linewidth_,label='Sim. $q_8$')
ax[1].plot(time_list,exp_exact_hp_switch_n9,linewidth=linewidth_,label='The. $q_9$')
# ax[1].plot(time_list,exp_exp_hp_switch_z9,'--o',linewidth=linewidth_,label='Sim. $q_9$')
ax[1].errorbar(time_list,mean_rl_z9,yerr=std_rl_z9,fmt='--o',linewidth=linewidth_,label='Sim. $q_9$')
# ax[1].set_ylabel('$\langle n \\rangle$',fontsize=fontsize_)
ax[1].set_title('$H_{\\text{p}}(t_L,t_R)$',fontsize=fontsize_,pad=12)
ax[1].set_xlabel('time $t$',fontsize=fontsize_)
ax[1].set_xticks([0,1,2,3,4])
ax[1].set_yticks([.5,0,0.5,1])
ax[1].tick_params(axis='both',labelsize=fontsize_,pad=10)
for index in ['left','right','top','bottom']:
    ax[1].spines[index].set_linewidth(linewidth_)
ax[1].legend(fontsize=fontsize_-5)

fig.text(0.04,0.92,r'\bfseries (a)',fontsize=fontsize_)
fig.text(0.52,0.92,r'\bfseries (b)',fontsize=fontsize_)

plt.tight_layout(w_pad=2)
plt.savefig('hn_2026.pdf')
plt.show()
```
